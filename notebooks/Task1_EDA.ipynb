```python
import os

print(os.listdir("/content/PlantVillage"))
```

```python
dataset_path = "/content/PlantVillage/PlantVillage"
```

```python
classes = sorted(os.listdir(dataset_path))

print("Total Classes:", len(classes))
print(classes)
```

```python
import pandas as pd

image_count = []

for cls in classes:
    folder = os.path.join(dataset_path, cls)
    count = len(os.listdir(folder))
    image_count.append([cls, count])

df = pd.DataFrame(image_count, columns=["Class", "Images"])

df
```

```python
df.to_csv("class_distribution.csv", index=False)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

sample_class = classes[0]

sample_folder = os.path.join(dataset_path, sample_class)

sample_image = os.listdir(sample_folder)[0]

image_path = os.path.join(sample_folder, sample_image)

img = Image.open(image_path)

plt.figure(figsize=(6,6))
plt.imshow(img)
plt.title(sample_class)
plt.axis("off")
```
